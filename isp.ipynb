```python
dados = 'https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv'
df = pd.read_csv(dados, encoding= 'latin1', sep = ';')
df
```

```python
# Agrupando por CISP e somando os furtos de celular
df_cisp = df.groupby('cisp')['furto_celular'].sum().reset_index()
# Ordenando do maior para o menor
df_cisp = df_cisp.sort_values(by='furto_celular', ascending=False)
df_cisp
```

```python
df_roubo_de_veiculo = df.groupby('cisp')['roubo_veiculo'].sum().reset_index().sort_values(by = 'roubo_veiculo', ascending=False).head(15)
df_roubo_de_veiculo
```

```python
#analisar apenas a cidade do Rio de Janeiro (capital)
# Usando .loc. Loc é um contexto de filtro
df_capital = df.loc[df['munic'] == 'Rio de Janeiro']
df
```

```python
df_maiores_qtde_furto_veiculo = df_capital.groupby('cisp')['furto_veiculos'].sum().reset_index().sort_values(by = 'furto_veiculos', ascending=False)
df_maiores_qtde_furto_veiculo
```

```python
array_cisp_capital_furto = np.array(df_maiores_qtde_furto_veiculo['furto_veiculos'])
array_cisp_capital_furto
```

```python
media = np.mean(array_cisp_capital_furto)
mediana = np.median(array_cisp_capital_furto)
q1 = np.percentile(array_cisp_capital_furto, 25)
q2 = np.percentile(array_cisp_capital_furto, 50)
q3 = np.percentile(array_cisp_capital_furto, 75)
distancia = (media - mediana) / mediana
print('Q1:', q1)
print('Q2:', q2)
print('Q3:', q3)
print(f'média {round(media,2)}\nMediana {mediana}') #"round" serve para arredondar o número!
print(f'distancia {distancia:.2%}')
```

```python
top_25 = df_maiores_qtde_furto_veiculo.loc[df_maiores_qtde_furto_veiculo['furto_veiculos'] > q3]
top_25
```

```python
df_roubo_transeunte = df.groupby('cisp')['roubo_transeunte'].sum().reset_index().sort_values(by = 'roubo_transeunte', ascending=False)
df_roubo_transeunte

```

```python
array_roubo_transe_capital = np.array(df_roubo_transeunte['roubo_transeunte'])
array_roubo_transe_capital
```

```python
media = np.mean(array_roubo_transe_capital)
mediana = np.median(array_roubo_transe_capital)
distancia = (media - mediana) / mediana
q1 = np.percentile(array_roubo_transe_capital,25)
q3 = np.percentile(array_roubo_transe_capital,75)
iqr = q3 - q1
limte_superior = q3 + (1.5 * iqr)
limte_inferior = q1 - (1.5 * iqr)
print(f'média {media}') 
print(f'mediana {mediana:.2f}')
print(f'distancia {distancia:.2%}')
print(f'q1: {q1}')
print(f'q3: {q3}')
print(f'Limite Superior {limte_superior}')
print(f'Limite Inferior {limte_inferior}')
```

```python
outlier_transeunte_cisp = df_roubo_transeunte.loc[df_roubo_transeunte['roubo_transeunte'] > limte_superior]
outlier_transeunte_cisp
```
