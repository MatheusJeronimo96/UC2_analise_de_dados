```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

### 1 Construção da base consolidada

```python
df_2024_1 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2024.01.csv', sep = ';' , decimal = ',')
df_2024_2 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2024.02.csv', sep = ';' , decimal = ',')
df_2025_1 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2025.01.csv', sep = ';' , decimal = ',')
df_2025_2 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2025.02.csv', sep = ';' , decimal = ',')
```

```python
df = pd.concat([df_2024_1,df_2024_2,df_2025_1,df_2025_2], ignore_index=True)
df.shape
df.head()
```

```python
df.columns
```

### 2 Gasolina C no Rio de Janeiro

```python
df_estado_rj = df.loc[df['Estado - Sigla'] == 'RJ']
df_estado_rj
```

```python
df_media_gasolina_c = df_estado_rj.loc[df['Produto'] == 'GASOLINA']
df_media_gasolina_c
```

```python
df_mdia_munic_venda = df_estado_rj.groupby('Municipio')['Valor de Venda'].mean().reset_index().sort_values(by = 'Valor de Venda', ascending=False)
df_mdia_munic_venda
```

```python
df_estado_rj['Produto'].unique()
```

### 3 Outliers de preço: Gasolina C no RJ


```python
Q1 = df_media_gasolina_c['Valor de Venda'].quantile(0.25)
Q3 = df_media_gasolina_c['Valor de Venda'].quantile(0.75)

IQR = Q3 - Q1
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print('Q1:', Q1)
print('Q3:', Q3)
print('IQR:', IQR)
print('Limite inferior: R$', limite_inferior)
print('Limite superior: R$', limite_superior)
```

```python

outliers_inferiores = df_media_gasolina_c[df_media_gasolina_c['Valor de Venda'] < limite_inferior].sort_values(by='Valor de Venda', ascending=True)
outliers_superiores = df_media_gasolina_c[df_media_gasolina_c['Valor de Venda'] > limite_superior].sort_values(by='Valor de Venda', ascending=False)

print('Quantidade de outliers inferiores:', len(outliers_inferiores))
print('Quantidade de outliers superiores:', len(outliers_superiores))
```

### 4 Comparando estados: Etanol Hidratado


```python
df_etanol_rj = df_estado_rj.loc[df['Produto'] == 'ETANOL']
df_etanol_rj 
```

```python
df_estado_sp = df.loc[df['Estado - Sigla'] == 'SP']
df_estado_sp
```

```python
df_etanol_sp = df_estado_sp.loc[df['Produto'] == 'ETANOL']
df_etanol_sp
```

```python
df_media_etanol_rj = df_etanol_rj['Valor de Venda'].mean()
df_media_etanol_sp = df_etanol_sp['Valor de Venda'].mean()

print(f'Media do etanol no Rj: R$ {df_media_etanol_rj:.2f}')
print(f'Media do etanol no SP: R$ {df_media_etanol_sp:.2f}')
```

```python
Q1_rj = df_etanol_rj['Valor de Venda'].quantile(0.25)
Q3_rj = df_etanol_rj['Valor de Venda'].quantile(0.75)
IQR_rj = Q3_rj - Q1_rj
limite_inferior_rj = Q1_rj - 1.5 * IQR_rj
limite_superior_rj = Q3_rj + 1.5 * IQR_rj


Q1_sp = df_etanol_sp['Valor de Venda'].quantile(0.25)
Q3_sp = df_etanol_sp['Valor de Venda'].quantile(0.75)
IQR_sp = Q3_sp - Q1_sp
limite_inferior_sp = Q1_sp - 1.5 * IQR_sp
limite_superior_sp = Q3_sp + 1.5 * IQR_sp

print(f'RJ — IQR: {IQR_rj:.2f} | limites: {limite_inferior_rj:.2f} a {limite_superior_rj:.2f}')
print(f'SP — IQR: {IQR_sp:.2f} | limites: {limite_inferior_sp:.2f} a {limite_superior_sp:.2f}')
```

```python
outliers_rj_inferiores = df_etanol_rj.loc[df_etanol_rj['Valor de Venda'] < limite_inferior_rj].sort_values(by='Valor de Venda', ascending=True)
outliers_rj_superiores = df_etanol_rj.loc[df_etanol_rj['Valor de Venda'] > limite_superior_rj].sort_values(by='Valor de Venda', ascending=False)
outliers_sp_inferiores = df_etanol_sp.loc[df_etanol_sp['Valor de Venda'] < limite_inferior_sp].sort_values(by='Valor de Venda', ascending=True)
outliers_sp_superiores = df_etanol_sp.loc[df_etanol_sp['Valor de Venda'] > limite_superior_sp].sort_values(by='Valor de Venda', ascending=False)

print('RJ — outliers inferiores:', len(outliers_rj_inferiores))
print('RJ — outliers superiores:', len(outliers_rj_superiores))
print('SP — outliers inferiores:', len(outliers_sp_inferiores))
print('SP — outliers superiores:', len(outliers_sp_superiores))
```

## 5 Bairros da capital: preços dentro ou fora do padrão?

```python
bairros_outliers_inferiores = df_media_por_bairro[df_media_por_bairro < limite_inferior_bairros].sort_values(ascending=True)
bairros_outliers_superiores = df_media_por_bairro[df_media_por_bairro > limite_superior_bairros].sort_values(ascending=False)
print('Bairros com médias inferiores atípicas:')
print(bairros_outliers_inferiores)
print('Bairros com médias superiores atípicas:')
print(bairros_outliers_superiores)
```
